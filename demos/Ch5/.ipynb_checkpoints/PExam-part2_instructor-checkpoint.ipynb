```r
library(dplyr)
library(nsprcomp)
```

```r
twitter <- read.csv("../../../hw/twitter.csv")
```

```r
twitter <- twitter %>% mutate_all(~(scale(.) %>% as.vector))
```

```r
cov_matrix <- cov(twitter)
e <- eigen(cov_matrix)$values
plot(e)


# Percentage of variance explained by the first 1, 2, 3, 5, 10, and 20 principal components:

PCA_results <- prcomp(twitter, scale = TRUE)
var_explained <- PCA_results$sdev^2 / sum(PCA_results$sdev^2)
rank <- c(1, 2, 3, 5, 10, 20)
for (x in rank) {
  total_var <- sum(var_explained[1:x])
  print(total_var * 100)  
}

svd_ <- svd(twitter)
err <- numeric(length(rank))

for(i in 1:length(err)){
    r_i <- rank[i]
    err[i] <- mean((as.matrix(twitter) - sweep(svd_$u[,1:r_i,drop = FALSE], FUN = "*", MARGIN = 2, STATS = svd_$d[1:r_i])  %*% t(svd_$v[,1:r_i,drop = FALSE]))^2)
}    


### c.
phi <- -(eigen(cov_matrix)$vectors[,1:2])
row.names(phi) <- colnames(twitter)
colnames(phi) <- c("PC1", "PC2")

PC1 <- as.matrix(twitter) %*% phi[,1]
PC2 <- as.matrix(twitter) %*% phi[,2]
plot(PC1, PC2)
```

```r
nnpca <- nsprcomp(twitter, 
         retx = TRUE, 
         ncomp = 10, 
         nneg = TRUE, 
         center = FALSE, 
         scale. = FALSE)
```

```r
str(nnpca)
```

```r
X_hat <- predict(nnpca)%*%MASS::ginv(nnpca$rotation)
E2 = ((X_hat - as.matrix(twitter))^2)
```

```r
mean(E2)
```

```r
str(nnpca$x)
```

```r
plot(nnpca$x[,1], nnpca$x[,2])
```

```r
# missing piece
err_nnpca <- numeric(length(rank))
counter <- 0
for(i in rank){
counter <- counter + 1
nnpca_ <- nsprcomp(twitter, 
         retx = TRUE, 
         ncomp = i, 
         nneg = TRUE, 
         center = FALSE, 
         scale. = FALSE)
X_hat_ <- predict(nnpca_)%*%MASS::ginv(nnpca_$rotation)
err_nnpca[counter] = mean(((X_hat_ - as.matrix(twitter))^2))
}   
```

```r
err_nnpca
```

```r
# reconstruction errors of NNPCA close to PCA.
plot(rank, err)
lines(rank, err_nnpca, lty = "dashed")
```
