```r
library(dplyr)
library(nsprcomp)
library(MASS)
```

```r
data(iris)
head(iris)
```

```r
set.seed(142)
Y <- iris$Species
irisX <- iris[,c("Sepal.Length", "Sepal.Width", "Petal.Length", "Petal.Width")] %>% mutate_all(~(scale(.) %>% as.vector))
dnoise <- 16
n <- length(Y)
X <- cbind(irisX, matrix(rnorm(n * dnoise, sd = 1), nrow = n, ncol = dnoise)) 
X <- X %>% mutate_all(~(scale(.) %>% as.vector))
```

```r
str(irisY)
```

```r
cov_matrix <- cov(X)
e <- eigen(cov_matrix)$values
plot(e)
```

```r
# Percentage of variance explained by the first 1, 2, 3, 5, 10, and 20 principal components:

PCA_results <- prcomp(X, scale = TRUE)
var_explained <- PCA_results$sdev^2 / sum(PCA_results$sdev^2)
rank <- 1:10
for (x in rank) {
  total_var <- sum(var_explained[1:x])
  print(total_var * 100)  
}

svd_ <- svd(X)
err <- numeric(length(rank))

for(i in 1:length(err)){
    r_i <- rank[i]
    err[i] <- mean((as.matrix(X) - sweep(svd_$u[,1:r_i,drop = FALSE], FUN = "*", MARGIN = 2, STATS = svd_$d[1:r_i])  %*% t(svd_$v[,1:r_i,drop = FALSE]))^2)
}    


### c.
phi <- -(eigen(cov_matrix)$vectors[,1:2])
row.names(phi) <- colnames(X)
colnames(phi) <- c("PC1", "PC2")

PC1 <- as.matrix(X) %*% phi[,1]
PC2 <- as.matrix(X) %*% phi[,2]
plot(PC1, PC2, col = as.numeric(Y)+1, pch = 16)
```

```r
err
```

```r
nnpca <- nsprcomp(t(X), 
         retx = TRUE, 
         ncomp = 3, 
         nneg = TRUE, 
         center = FALSE, 
         scale. = FALSE)
```

```r
str(nnpca)
```

```r
Ahat <- nnpca$rotation  
```

```r
Xhat <- t(predict(nnpca)%*%ginv(nnpca$rotation))
```

```r
err_nnpca3 <- mean((Xhat - as.matrix(X))^2)
print(err_nnpca3)
```

```r
str(Ahat)
```

```r
par(mfrow = c(2,2))
plot(Ahat[,1], Ahat[,2], col = as.numeric(Y)+1, pch = 16)
plot(Ahat[,1], Ahat[,3], col = as.numeric(Y)+1, pch = 16)
plot(Ahat[,2], Ahat[,3], col = as.numeric(Y)+1, pch = 16)
```

```r
ldares <- lda(Y ~ as.matrix(X), proportion = rep(1/3, 3))
```

```r
lda_pred <- predict(ldares, newdata = X, dimen = 2)
```

```r
str(lda_pred)
```

```r
plot(lda_pred$x[,1], lda_pred$x[,2], col = as.numeric(Y)+1, pch = 16)
```
